```python
# DESIGN DE SOFTWARE 2 - EXERCÍCIO DE LISTAS
# JOÃO PEDRO ALVES OLIVEIRA
# ===========================================
# Crie um programa em Phyton que permita ao usuário informar as notas de 5 alunos e armazende essas notas em uma lista.
# Ao final, o programa deverá:
# Mostrar todas as notas cadastradas.
# Calcular e mostrar a média da turma.
# Mostrar a maior nota.
# Mostrar a menor nota.
# Informar quantos alunos foram aprovados (nota >= 6).
# Informar quantos alunos foram reprovados (nota < 6).
```

```python
# definindo as variáveis
notas = []
media = 0
apr = 0
rpr = 0
# quantidade de alunos
alunos = 5

# zerando a variável
i = 0
# adicionando as notas na lista e na média
while i < alunos:
  print("Digite a " + str(i+1) + "º nota: ")
  notas.insert(i, float(input()))
  media += notas[i]
  i += 1

# calculando da média
media /= alunos
notas.sort()
# exibindo as notas, média e menor e maior nota
print("\nAs notas foram: " + str(notas))
print("A média da turma é: " + str(media))
print("A maior nota foi: " + str(notas[0]))
print("A menor nota foi: " + str(notas[alunos - 1]))

# zerando a variável
i = 0
# conferindo quantidade de alunos aprovados e reprovados
while i < alunos:
  if notas[i] < 6 :
    rpr += 1
  else:
    apr += 1
  i += 1

# exibindo quantos alunos foram aprovados e reprovados
print(str(apr) + " Alunos foram aprovados")
print(str(rpr) + " Alunos foram reprovados")
```
