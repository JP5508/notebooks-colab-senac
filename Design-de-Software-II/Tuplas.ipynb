```python
# DESIGN DE SOFTWARE 2 - EXERCÍCIOS DE TUPLAS
# JOÃO PEDRO ALVES OLIVEIRA
# ===========================================
# EXERCÍCIO 1
# Crie um programa que:
# Armazene quatro notas em uma tupla.
# Exiba todas as notas.
# Calcule e exiba a média das notas.
# Mostre a maior e a menor nota.
# Verifique se o aluno foi aprovado, considerando a média mínima igual a 6,0.
```

```python
# definindo as variáveis
notas = (4.5, 7, 8, 6)
ordem = []
media = 0

# calculando a média
i = 0
while i < 4:
    media += notas[i]
    ordem.insert(i, notas[i])
    i += 1
media /= 4
ordem.sort()

# exibindo informações
print("As notas foram: " + str(notas))
print("A média do aluno é: " + str(media))
print("A maior nota foi: " + str(ordem[0]))
print("A menor nota foi: " + str(ordem[3]))

# aprovado ou reprovado
if media < 6:
    print("O aluno foi reprovado")
else:
    print("O aluno foi aprovado")
```
