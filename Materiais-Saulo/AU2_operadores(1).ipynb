## **1. Operadores Aritméticos**

Operadores aritméticos são símbolos que usamos para fazer cálculos matemáticos.

| Operador | Símbolo | Exemplo | Resultado |
|----------|---------|---------|-----------|
| Soma | `+` | `5 + 3` | `8` |
| Subtração | `-` | `10 - 4` | `6` |
| Multiplicação | `*` | `3 * 4` | `12` |
| Divisão | `/` | `15 / 3` | `5.0` |
| Divisão inteira | `//` | `15 // 4` | `3` |
| Resto (módulo) | `%` | `15 % 4` | `3` |
| Potência | `**` | `2 ** 3` | `8` |

## **2. Operadores Relacionais**

Operadores relacionais (também chamados de **comparação**) são usados para comparar valores. Eles sempre retornam `True` ou `False`.

| Operador | Significado | Exemplo | Resultado |
|----------|-------------|---------|-----------|
| `==` | Igual a | `5 == 5` | `True` |
| `!=` | Diferente de | `5 != 3` | `True` |
| `>` | Maior que | `10 > 5` | `True` |
| `<` | Menor que | `3 < 7` | `True` |
| `>=` | Maior ou igual | `5 >= 5` | `True` |
| `<=` | Menor ou igual | `4 <= 6` | `True` |

## **3. Operadores Lógicos**

| `and` | E (ambas condições verdadeiras) | `True and False` | `False` |
| `or` | OU (pelo menos uma verdadeira) | `True or False` | `True` |
| `not` | NÃO (inverte o valor) | `not True` | `False` |
