```python
#ex1 - par ou impar

#pede o número ao usuário
print('Esse programa vai informar se o número informado é par ou impar.')
print('Digite um número:')
user_num = int(input())

#verifica se é par ou impar
i = user_num

if i == 0:
    print(str(i) + ' é par')
else:
    #subtrai 2 até chegar em 0 ou 1
    while i >= 2:
        i = i - 2

    #se sobrou 0 é par
    if i == 0:
        print(str(user_num) + ' é par')
    #se sobrou 1 é ímpar
    else:
        print(str(user_num) + ' é ímpar')
```

```python
#ex2 - gratuidade por idade

#perguntando a idade do usuário
print('Esse programa vai informar se o usuário tem gratuidade no transporte baseado na sua idade')
print('Digite sua idade:')
user_age = int(input())

#verificar idade
if user_age < 65:
 print('Não tem gratuidade no transporte')
elif user_age >= 65:
 print('Tem gratuidade no transporte')
else:
 print('Digite valores válidos')
```

```python
#ex3 - resultado de aprovação do aluno

#perguntando as notas do aluno
print('Esse programa vai tirar a média das duas notas do aluno para informar se ele foi aprovado')
print('Digite a primeira nota:')
user_nota1 = float(input())
print('Digite a segunda nota:')
user_nota2 = float(input())

#tira a média das notas
aluno_media = (user_nota1 + user_nota2) / 2

if aluno_media < 5:
 print('O aluno foi reprovado com média ' + str(aluno_media))
elif aluno_media > 7:
 print('O aluno foi aprovado com média ' + str(aluno_media))
else:
 print('O aluno ficou de recuperação com média ' + str(aluno_media))
```

```python
#ex4 - desconto da ótica

#pergunta a idade do usuário
print('Digite a idade do usuário:')
user_idade = int(input())

#definindo o desconto
if user_idade <= 10:
  otica_desconto = 10
elif user_idade >= 80:
  otica_desconto = 80
else:
  otica_desconto = user_idade

#calculando e exibindo o valor com desconto
valor_final = 200.00 - ((200 * otica_desconto) / 100)

print('O valor final com o desconto de ' + str(otica_desconto) + '% é: R$' + str(valor_final))
```

```python
#ex5 - reajuste salarial

#perguntando o salário ao usuário
print('Qual seu salário atual?')
user_sal_atual = float(input())

#definindo aumento
if user_sal_atual < 1500:
  porc_reajuste = 25
elif user_sal_atual >= 1500 and user_sal_atual < 2000:
  porc_reajuste = 20
elif user_sal_atual >= 2000 and user_sal_atual < 3000:
  porc_reajuste = 15
elif user_sal_atual >= 3000 and user_sal_atual < 5000:
  porc_reajuste = 10
else:
  porc_reajuste = 5

#calculando e exibindo os resultados
val_reajuste = ((user_sal_atual * porc_reajuste) / 100)
sal_final = user_sal_atual + val_reajuste

print('O salário atual é: R$' + str(user_sal_atual))
print('A percentagem do reajuste é de: ' + str(porc_reajuste) + '%')
print('O aumento foi de: R$' + str(val_reajuste))
print('O salario após o reajuste será: R$' + str(sal_final))
```

```python
#ex6 - jokenpô

#1-pedra|2-papel|3-tesoura
#importação do random
import random as rd

print('Digite "pedra", "papel", "tesoura" ou "Q" para parar de jogar')

#loop do jogo
while True:
    pc_escolha = rd.randint(1, 3)
    user_escolha = str(input())


    #caso digite Q o jogo e finalizado
    if user_escolha.lower() == 'q':
        print('Jogo Finalizado')
        break


    if user_escolha.lower() == 'pedra':
        if pc_escolha == 1:
            print('Empate! Ambos escolheram pedra')
        elif pc_escolha == 2:
            print('Você perdeu! Papel cobre pedra')
        else:
            print('Você ganhou! Pedra quebra tesoura')


    elif user_escolha.lower() == 'papel':
        if pc_escolha == 1:
            print('Você ganhou! Papel cobre pedra')
        elif pc_escolha == 2:
            print('Empate! Ambos escolheram papel')
        else:
            print('Você perdeu! Tesoura corta papel')


    elif user_escolha.lower() == 'tesoura':
        if pc_escolha == 1:
            print('Você perdeu! Pedra quebra tesoura')
        elif pc_escolha == 2:
            print('Você ganhou! Tesoura corta papel')
        else:
            print('Empate! Ambos escolheram tesoura')


    else:
        print('Digite valores válidos (pedra, papel, tesoura ou Q)')
```
