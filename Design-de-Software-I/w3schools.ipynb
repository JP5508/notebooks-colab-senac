```python
#teste

print('O que você quer calcular?')
print('Tensão, corrente ou resistência')
user_calc = input()

if user_calc.lower() == 'tensão' or 'tensao':

   print('\n Calculo de TENSÃO \n')

   print('Qual é a resistência?')
   user_resi = float(input())
   print('Qual é a corrente?')
   user_corr = float(input())

   resp_tens = user_resi * user_corr
   print('A tensão é ' + str(resp_tens) + 'V')

elif user_calc.lower() == 'corrente':

   print('\n Calculo de CORRENTE \n')

   print('Qual é a tensão?')
   user_tens = float(input())
   print('Qual é a resistência?')
   user_resi = float(input())

   resp_corr = user_tens / user_resi
   print('A corrente é ' + str(resp_corr) + 'A')

elif user_calc.lower() == 'resistência' or 'resistencia':

   print('\n Calculo de RESISTÊNCIA \n')

   print('Qual é a tensão?')
   user_tens = float(input())
   print('Qual é a corrente?')
   user_corr = float(input())

   resp_resi = user_tens / user_corr
   print('A tensão é ' + str(resp_resi) + 'Ω')

else:
   print('Digite valores válidos')
```
